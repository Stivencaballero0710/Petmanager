## 1. Análisis de la situación — Araujo (1:30)

Una veterinaria necesita relacionar de forma coherente los registros de clientes, animales, personal, atenciones clínicas, inventario y facturación. PetManager se plantea como una base de datos relacional para organizar y recuperar esta información sin repetir datos que pertenecen a otras entidades.

**Mostrar y explicar:** problema, contexto, objetivo y alcance.

## 3. Modelo relacional y normalización — Caballero (3:00)

El modelo inicial se construyó con **18 tablas** y sus dominios y restricciones PK, FK, UK y NN. La revisión documentada conserva las 18 tablas de **1FN a 5FN, incluida BCNF**; en **6FN** el ejercicio descompone la estructura hasta obtener **54 tablas**.

| Etapa | Resultado |
|---|---:|
| 1FN: valores atómicos | 18 tablas |
| 2FN: sin dependencias parciales | 18 |
| 3FN: control de dependencias | 18 |
| BCNF: determinantes como superclaves | 18 |
| 4FN: revisión de multivaluadas | 18 |
| 5FN: dependencias de reunión | 18 |
| 6FN: descomposición de atributos | **54** |

**Ejemplo principal:** CLIENTE pasa de siete atributos a seis relaciones en 6FN (CLIENTE y cinco proyecciones separadas). El informe propone **implementar 18 tablas en 5FN** para evitar un número elevado de JOIN.
